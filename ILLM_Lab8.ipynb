```python
#Load, clean, and inspect the dataset
import subprocess

code = r'''
from huggingface_hub import hf_hub_download
import pandas as pd
import re

path = hf_hub_download(
    repo_id="ai4bharat/BPCC",
    filename="ilci/guj_Gujr.tsv",
    repo_type="dataset"
)

df = pd.read_csv(path, sep="\t", encoding="utf-8")

print("Original dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# Keep only English-to-Gujarati pairs
df = df[
    (df["src_lang"] == "eng_Latn") &
    (df["tgt_lang"] == "guj_Gujr")
][["src", "tgt"]].copy()

# Remove missing values and normalize whitespace
df = df.dropna(subset=["src", "tgt"])
df["src"] = df["src"].astype(str).str.replace(r"\s+", " ", regex=True).str.strip()
df["tgt"] = df["tgt"].astype(str).str.replace(r"\s+", " ", regex=True).str.strip()

# Remove empty sentences and duplicate pairs
df = df[(df["src"] != "") & (df["tgt"] != "")]
df = df.drop_duplicates(subset=["src", "tgt"]).reset_index(drop=True)

# Count English whitespace-separated words
df["src_length"] = df["src"].apply(lambda s: len(s.split()))

# Four exhaustive, non-overlapping bins:
# Bin 1: <= 10 words
# Bin 2: 11–20 words
# Bin 3: 21–29 words
# Bin 4: >= 30 words
def assign_bin(length):
    if length <= 10:
        return "Bin 1 (<=10)"
    elif length <= 20:
        return "Bin 2 (11-20)"
    elif length < 30:
        return "Bin 3 (21-29)"
    else:
        return "Bin 4 (>=30)"

df["length_bin"] = df["src_length"].apply(assign_bin)

print("\nCleaned dataset shape:", df.shape)
print("\nBin distribution:")
print(df["length_bin"].value_counts().sort_index())

print("\nExamples:")
print(df[["src", "tgt", "src_length", "length_bin"]].head(5).to_string(index=False))
'''

result = subprocess.run(
    ["/content/onmt-env/bin/python", "-c", code],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print("STDERR:\n", result.stderr[-3000:])
```

```python
#Create length-aware train, validation, and test splits

import subprocess

code = r'''
from huggingface_hub import hf_hub_download
import pandas as pd
from sklearn.model_selection import train_test_split
from pathlib import Path

path = hf_hub_download(
    repo_id="ai4bharat/BPCC",
    filename="ilci/guj_Gujr.tsv",
    repo_type="dataset"
)

df = pd.read_csv(path, sep="\t", encoding="utf-8")
df = df[
    (df["src_lang"] == "eng_Latn") &
    (df["tgt_lang"] == "guj_Gujr")
][["src", "tgt"]].dropna().copy()

for col in ["src", "tgt"]:
    df[col] = df[col].astype(str).str.replace(r"\s+", " ", regex=True).str.strip()

df = df[(df["src"] != "") & (df["tgt"] != "")]
df = df.drop_duplicates(subset=["src", "tgt"]).reset_index(drop=True)
df["src_length"] = df["src"].apply(lambda s: len(s.split()))

def assign_bin(n):
    if n <= 10:
        return "Bin 1 (<=10)"
    elif n <= 20:
        return "Bin 2 (11-20)"
    elif n < 30:
        return "Bin 3 (21-29)"
    return "Bin 4 (>=30)"

df["length_bin"] = df["src_length"].apply(assign_bin)

train_parts, val_parts, test_parts = [], [], []

for bin_name, group in df.groupby("length_bin", sort=True):
    if len(group) < 10:
        raise ValueError(f"{bin_name} has only {len(group)} rows; too few for this split.")

    train, temp = train_test_split(
        group, test_size=0.20, random_state=42
    )
    val, test = train_test_split(
        temp, test_size=0.50, random_state=42
    )

    train_parts.append(train)
    val_parts.append(val)
    test_parts.append(test)

train_df = pd.concat(train_parts).sample(frac=1, random_state=42).reset_index(drop=True)
val_df = pd.concat(val_parts).sample(frac=1, random_state=42).reset_index(drop=True)
test_df = pd.concat(test_parts).sample(frac=1, random_state=42).reset_index(drop=True)

out = Path("/content/ilci_gujarati")
out.mkdir(parents=True, exist_ok=True)

for name, data in [("train", train_df), ("valid", val_df), ("test", test_df)]:
    data[["src", "tgt"]].to_csv(
        out / f"{name}.tsv", sep="\t", index=False, header=False
    )
    data[["src"]].to_csv(out / f"{name}.src", index=False, header=False)
    data[["tgt"]].to_csv(out / f"{name}.tgt", index=False, header=False)

print("Split sizes:")
print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nCounts by bin:")
for name, data in [("Train", train_df), ("Validation", val_df), ("Test", test_df)]:
    print(f"\n{name}")
    print(data["length_bin"].value_counts().sort_index())

print("\nFiles saved in:", out)
'''

result = subprocess.run(
    ["/content/onmt-env/bin/python", "-c", code],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print("STDERR:\n", result.stderr[-3000:])
```

```python
#Prepare a small training subset

import subprocess

code = r'''
import pandas as pd
from pathlib import Path

base = Path("/content/ilci_gujarati")
train = pd.read_csv(
    base / "train.tsv",
    sep="\t",
    header=None,
    names=["src", "tgt"]
)

train["length"] = train["src"].astype(str).apply(
    lambda x: len(x.split())
)

def assign_bin(n):
    if n <= 10:
        return "bin1"
    elif n <= 20:
        return "bin2"
    elif n < 30:
        return "bin3"
    return "bin4"

train["bin"] = train["length"].apply(assign_bin)

parts = []
for bin_name, group in train.groupby("bin"):
    sample = group.sample(
        n=min(2000, len(group)),
        random_state=42
    )
    parts.append(sample)

small_train = pd.concat(parts).sample(
    frac=1, random_state=42
).reset_index(drop=True)

small_train[["src"]].to_csv(
    base / "train_small.src", index=False, header=False
)
small_train[["tgt"]].to_csv(
    base / "train_small.tgt", index=False, header=False
)

print("Training examples:", len(small_train))
print("\nExamples per bin:")
print(small_train["bin"].value_counts().sort_index())
print("\nTraining files saved.")
'''

result = subprocess.run(
    ["/content/onmt-env/bin/python", "-c", code],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print(result.stderr[-3000:])
```

```python

# Translate the test set and evaluate BLEU scores

import subprocess
from pathlib import Path

base = Path("/content/ilci_gujarati")
model = base / "guj_transformer_step_2000.pt"
source_file = base / "test.src"
prediction_file = base / "test_predictions.txt"

# Check that the required files exist
for file in [model, source_file, base / "test.tsv"]:
    if not file.exists():
        raise FileNotFoundError(f"Required file not found: {file}")

# Translate English sentences into Gujarati
cmd = [
    "/content/onmt-env/bin/onmt_translate",
    "-model", str(model),
    "-src", str(source_file),
    "-output", str(prediction_file),
    "-gpu", "0",
    "-batch_size", "32",
    "-beam_size", "5",
    "-max_length", "100"
]

print("Translating test sentences. This may take a few minutes...")
result = subprocess.run(cmd, text=True, capture_output=True)

print(result.stdout[-4000:])
if result.stderr:
    print(result.stderr[-2000:])

if result.returncode != 0:
    raise RuntimeError("Translation failed. Check the error output above.")

print("Translation completed.")
print("Predictions saved to:", prediction_file)
```

```python

# Evaluate the translations using sacreBLEU

import subprocess

evaluation_code = r'''
import pandas as pd
from pathlib import Path
from sacrebleu import corpus_bleu

base = Path("/content/ilci_gujarati")

# Test TSV was saved without a header
test = pd.read_csv(
    base / "test.tsv",
    sep="\t",
    header=None,
    names=["src", "tgt"]
)

with open(base / "test_predictions.txt", encoding="utf-8") as f:
    predictions = [line.rstrip("\n") for line in f]

if len(test) != len(predictions):
    raise ValueError(
        f"Mismatch: {len(test)} test pairs but "
        f"{len(predictions)} predictions."
    )

test["prediction"] = predictions
test["src_length"] = test["src"].astype(str).apply(
    lambda s: len(s.split())
)

def assign_bin(n):
    if n <= 10:
        return "Bin 1 (<=10)"
    elif n <= 20:
        return "Bin 2 (11-20)"
    elif n < 30:
        return "Bin 3 (21-29)"
    return "Bin 4 (>=30)"

test["length_bin"] = test["src_length"].apply(assign_bin)

# Overall BLEU
overall = corpus_bleu(
    test["prediction"].tolist(),
    [test["tgt"].tolist()]
)

print("=" * 55)
print("OVERALL TEST RESULTS")
print("=" * 55)
print("Test sentences:", len(test))
print(f"Overall BLEU: {overall.score:.2f}")

# BLEU for each source sentence-length bin
print("\n" + "=" * 55)
print("BLEU BY SOURCE SENTENCE LENGTH")
print("=" * 55)

results = []

for bin_name in [
    "Bin 1 (<=10)",
    "Bin 2 (11-20)",
    "Bin 3 (21-29)",
    "Bin 4 (>=30)"
]:
    subset = test[test["length_bin"] == bin_name]

    if len(subset) == 0:
        print(f"{bin_name}: no examples")
        continue

    score = corpus_bleu(
        subset["prediction"].tolist(),
        [subset["tgt"].tolist()]
    )

    results.append({
        "Length Bin": bin_name,
        "Test Sentences": len(subset),
        "BLEU": round(score.score, 2)
    })

results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))

results_df.to_csv(base / "bin_wise_bleu.csv", index=False)
print("\nResults saved to:", base / "bin_wise_bleu.csv")
'''

result = subprocess.run(
    ["/content/onmt-env/bin/python", "-c", evaluation_code],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr[-4000:])
    raise RuntimeError("BLEU evaluation failed.")
```
